```python
import mne
import pandas as pd
import os

# Enable interactive plotting with Qt
%matplotlib qt

# Define the subjects to analyze
subjects_to_analyze = ['S5', 'S6', 'S7', 'S8', 'S9', 'S10', 'S11', 'S12', 'S13', 'S14', 'S15', 'S16', 'S17', 'S18', 'S19', 'S20', 'S21', 'S22', 'S23', 'S24']  # Add your subjects here
#subjects_to_analyze = ['S3']
# Define the file to visualize (Change this as needed)
file_to_visualize = "S19-pr-annotated_raw.fif"  # Example annotated file for visualization

# Define the path to the Excel sheet (same folder as code)
timeframes_file = "TimeFrames_Epoch.xlsx"

# Load the Excel data
timeframes_data = pd.read_excel(timeframes_file)

# Process each subject
for subject in subjects_to_analyze:
    print(f"Processing subject: {subject}")

    # Define file names for pr and po .snirf files
    pr_file = f"{subject}-pr.snirf"
    po_file = f"{subject}-po.snirf"

    # Check if the files exist
    if not os.path.exists(pr_file) or not os.path.exists(po_file):
        print(f"Missing files for subject {subject}: {pr_file} or {po_file}")
        continue

    # Load the .snirf files
    pr_raw = mne.io.read_raw_snirf(pr_file, preload=True)
    po_raw = mne.io.read_raw_snirf(po_file, preload=True)

    # Extract the start times of the recordings (epoch time in seconds)
    pr_start_time = pr_raw.info["meas_date"].timestamp()
    po_start_time = po_raw.info["meas_date"].timestamp()

    # Adjust for time zone difference (add 5 hours)
    pr_start_time += 21600 # 5 hours in seconds 18000 for S1-S4, 21600 for S5-S24
    po_start_time += 21600  # 5 hours in seconds 18000 for S1-S4, 21600 for S5-S24

    # Filter rows for the current subject
    subject_data = timeframes_data[timeframes_data["Subject"] == subject]

    # Add annotations for each row in the Excel sheet
    for _, row in subject_data.iterrows():
        # Convert Excel times from milliseconds to seconds
        start_time = (row["Start_timeFrame"] / 1000)
        end_time = (row["End_timeFrame"] / 1000)
        duration = end_time - start_time

        # Define the description
        if row["Activity"] == "Baseline":
            # Always add Baseline annotations to the pr file
            description = "Baseline"
            pr_raw.annotations.append(onset=start_time - pr_start_time, duration=duration, description=description)
        else:
            # Skip rows without Fatigue unless they are Baseline
            fatigue = row["Fatigue"]
            if pd.isna(fatigue):
                continue  # Skip rows with no Fatigue and not Baseline

            activity = str(row["Activity"])
            stimulation = str(row["Stimulation"]) if not pd.isna(row["Stimulation"]) else "?"
            trial = row["Trial"] if not pd.isna(row["Trial"]) else "T?"
            description = f"{fatigue}-{activity}-{stimulation}-{trial}"

            # Add annotation to the appropriate file
            if fatigue == "pr":
                pr_raw.annotations.append(onset=start_time - pr_start_time, duration=duration, description=description)
            elif fatigue == "po":
                po_raw.annotations.append(onset=start_time - po_start_time, duration=duration, description=description)

    # Save the annotated files as .fif
    pr_annotated_file = f"{subject}-pr-annotated_raw.fif"
    po_annotated_file = f"{subject}-po-annotated_raw.fif"
    pr_raw.save(pr_annotated_file, overwrite=True)
    po_raw.save(po_annotated_file, overwrite=True)

    print(f"Annotated files saved: {pr_annotated_file}, {po_annotated_file}")


# Check if the file exists and visualize it
if os.path.exists(file_to_visualize):
    raw = mne.io.read_raw_fif(file_to_visualize, preload=True)
    raw.plot(duration=10, start=0, block=True)
else:
    print(f"File {file_to_visualize} does not exist.")
```
